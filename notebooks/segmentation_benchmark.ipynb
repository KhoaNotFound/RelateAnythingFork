```python
from pathlib import Path
import subprocess

repo = Path("/kaggle/working/RelateAnythingFork")
# Stop immediately if local edits prevent pulling the benchmark code.
subprocess.run(["git", "pull", "--ff-only", "origin", "test"], cwd=repo, check=True)
subprocess.run(["uv", "sync", "--extra", "benchmark"], cwd=repo, check=True)
```

```python
from pathlib import Path
import subprocess
from IPython.display import HTML, display

repo = Path("/kaggle/working/RelateAnythingFork")
output = Path("/kaggle/working/segmentation_benchmark")
pointer = output / "latest_report.txt"
previous = pointer.read_text().strip() if pointer.exists() else None
command = ["uv", "run", "--extra", "benchmark", "demo.py", "--benchmark", "--output-dir", str(output)]
completed = subprocess.run(command, cwd=repo)
current = pointer.read_text().strip() if pointer.exists() else None
if current is not None and current != previous and Path(current).is_file():
    display(HTML(filename=current))  # Also display fresh partial reports.
completed.check_returncode()  # Surface the original failure, not FileNotFoundError.
if current is None or current == previous or not Path(current).is_file():
    raise RuntimeError("No new benchmark report was generated; inspect the command output above.")
```
